```python
# 1️⃣ تنزيل المشروع والمكتبات المطلوبة
!git clone https://github.com/keatonhalvorson/apex-mega22-bot.git
%cd apex-mega22-bot
!pip install -q fastapi uvicorn websockets aiohttp requests pandas numpy psutil

# 2️⃣ تنزيل أداة Cloudflare Tunnel لتوفير رابط عالمي مجاني
!wget -q -nc https://github.com/cloudflare/cloudflared/releases/latest/download/cloudflared-linux-amd64.deb
!dpkg -i cloudflared-linux-amd64.deb > /dev/null 2>&1
print("✅ تم تجهيز البيئة بنجاح!")
```

```python
# 3️⃣ تشغيل البوت والداشبورد واستخراج الرابط الخارجي المباشر
import subprocess, time, re

print("🚀 تشغيل محرك APEX Sovereign ورادار الـ 51 عملة...")
p_bot = subprocess.Popen(["python3", "run_mega22_live.py", "--port", "8080", "--capital", "50.0", "--slots", "1"])
time.sleep(3)

print("🌐 إنشاء الرابط الخارجي الآمن (Cloudflare Tunnel)...")
p_tunnel = subprocess.Popen(["cloudflared", "tunnel", "--url", "http://localhost:8080"], stdout=subprocess.PIPE, stderr=subprocess.STDOUT, text=True)

tunnel_url = None
for _ in range(30):
    line = p_tunnel.stdout.readline()
    if "trycloudflare.com" in line:
        match = re.search(r"https://[a-zA-Z0-9-]+\.trycloudflare\.com", line)
        if match:
            tunnel_url = match.group(0)
            break
    time.sleep(0.5)

if tunnel_url:
    print("
" + "="*65)
    print("🎉 رابط الداشبورد المباشر الخاص بك (افتحه من هاتفك أو حاسوبك):")
    print(f"👉 {tunnel_url}")
    print("="*65 + "
")
else:
    print("❌ تعذر استخراج الرابط، يرجى إعادة تشغيل الخلية.")

# إبقاء البوت شغالاً
p_bot.wait()
```
